**Theory:**

Data wrangling, often referred to as data munging, involves the critical tasks of cleansing, reshaping, and organizing raw data into a coherent and analytical-ready structure. Given that real-world datasets frequently suffer from incompleteness, inconsistencies, or lack of structure, data wrangling is a fundamental prerequisite for any meaningful data analysis or model construction.

This assignment illustrates these concepts by generating a sample Pandas DataFrame with intentional gaps in 'Name', 'Age', and 'Score' columns. The `isnull()` function was employed to pinpoint these missing entries, guiding the subsequent cleaning actions. Specifically, missing 'Age' values were imputed using the column's mean via `fillna()`, preserving data integrity. Rows lacking 'Name' entries were systematically removed using `dropna()` to ensure dataset uniformity.

Beyond mere cleaning, data wrangling encompasses data transformation. An example of this was the creation of a 'Status' column, derived conditionally from the 'Score' values. This process of generating insightful new features from existing ones is termed column transformation, a vital aspect of feature engineering.

Furthermore, the assignment touches upon Exploratory Data Analysis (EDA), emphasizing its significance in comprehending data distributions, uncovering patterns, identifying anomalies, and facilitating informed decision-making. The discussion also briefly introduces Pandas' core data structures: Series (one-dimensional), DataFrame (two-dimensional), and the largely phased-out Panel (three-dimensional).

**5. Column Transformation:**

Column transformation refers to the operation of altering existing columns or deriving entirely new columns from the data already present within a dataset.

1.  This involves either modifying the values inside current columns or producing novel columns.
2.  It aids in feature engineering by constructing more informative variables for analysis.
3.  By providing a richer representation of the underlying problem, it can enhance predictive model efficacy.
4.  Examples include scaling numerical data, encoding categorical attributes, performing mathematical calculations, or applying custom functions.
5.  Its utility lies in converting unrefined data into a format better suited for deep analysis or machine learning applications.

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame with missing values for demonstration
data = {
    'Name': ['Sarang', 'Amit', 'Rohit', 'Neha', None],
    'Age': [20, 22, np.nan, 21, 23],
    'Score': [85, np.nan, 90, 88, 75]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)


print("\nIs Null Check:")
print(df.isnull().sum())

# 4d: Fill missing Age with the mean Age (Data Wrangling step)
df['Age'] = df['Age'].fillna(df['Age'].mean())

# 4c: Drop rows where Name is missing
df = df.dropna(subset=['Name'])

# 5: Column Transformation (Example: Adding a 'Status' column)
# If score > 80, Status is 'Pass', else 'Average'
df['Status'] = df['Score'].apply(lambda x: 'Pass' if x > 80 else 'Average')

print("\nWrangled DataFrame:")
print(df)
```

**1. What is data wrangling?**

Data wrangling, also known as data munging, is the critical process of converting raw, often untidy data into a cleaner, more structured, and accessible format, thereby enhancing its utility for analytical purposes.

**Key Aspects of Data Wrangling:**

1.  Encompasses the comprehensive activities of cleaning, structuring, and transforming unrefined data.
2.  Aims to achieve data uniformity, utility, and proper organization.
3.  Involves operations such as managing missing entries, eliminating redundant records, combining disparate datasets, and rectifying data formats.
4.  Its primary objective is to make data readily available and suitable for examination.

**Distinction from Data Preprocessing:**

1.  Data preprocessing is a subsequent, more specialized phase typically performed after wrangling.
2.  Its focus is on tailoring data specifically for machine learning algorithms.
3.  Techniques include data scaling, normalization, converting categorical data, and feature selection.
4.  The main purpose is to adapt data into an algorithm-compatible format.

**3. Main Differences: Series, DataFrame, and Panel:**

  Series: A one-dimensional, labeled array designed to store various data types.

  DataFrame: A two-dimensional, labeled tabular data structure where columns can have diverse data types, akin to a relational table or spreadsheet.

  Panel: A three-dimensional data container, which is now largely superseded in contemporary Pandas by the use of MultiIndex DataFrames.

**4. Explain the use of:**

  a. `isnull()`: Identifies the presence of null or missing data points; it generates a boolean structure mirroring the original data's dimensions.

  b. `notnull()`: Serves as the inverse of `isnull()`; it yields `True` for existing, non-missing values.

  c. `dropna()`: Eliminates rows or columns that contain any missing data from the dataset.

  d. `fillna()`: Populates missing data entries with either a user-specified constant or an aggregated value, such as an average.
