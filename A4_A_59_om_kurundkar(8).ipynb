**Theory**

Correlation and regression are fundamental statistical techniques used in data analysis to understand relationships between variables and make predictions. Correlation helps in identifying how strongly variables are related, while regression is used to model this relationship and predict outcomes. These concepts are widely used in fields like data science, machine learning, economics, and engineering for decision-making and analysis.

1. **Correlation :**
Correlation is a statistical measure that indicates the strength and direction of the relationship between two variables.
    
      * Library Used: NumPy, SciPy

      * Functions: np.corrcoef(), scipy.stats.pearsonr()



2. **Regression : **
Regression is a statistical method used to model and predict the relationship between a dependent and independent variable.

      * Library Used: Scikit-learn

      * Functions: LinearRegression(), .fit(), .predict()

3. **Karl Pearson’s Correlation Coefficient :**
Karl Pearson’s coefficient measures linear correlation between two variables, with values ranging from -1 to +1.

      * Library Used: NumPy, SciPy

      * Functions: np.corrcoef(), pearsonr()

4. **Simple Linear Regression :**
Simple linear regression establishes a linear relationship between one independent variable and one dependent variable using a straight line.

      * Library Used: Scikit-learn
     
      * Functions: LinearRegression()

5. **Regression Equation :**
The regression equation (y = mx + c) represents the mathematical relationship between variables used for prediction.

     * Library Used: Scikit-learn

     * Functions: .coef_, .intercept_

6. **Mean Squared Error (MSE) :**
Mean Squared Error measures the average squared difference between actual and predicted values.

      * Library Used: Scikit-learn

      * Functions: mean_squared_error()

7. **R² Score (Coefficient of Determination) :**
R² score represents how well the regression model explains the variance in the dependent variable.

      * Library Used: Scikit-learn

      * Functions: r2_score()

8. **Multicollinearity :**
Multicollinearity occurs when independent variables are highly correlated with each other in a regression model.

      * Library Used: Statsmodels (optional for detection)

      * Functions: variance_inflation_factor() (VIF)

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score


x = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([2, 5, 7, 9, 11, 13])

corr_matrix = np.corrcoef(x.flatten(), y)
karl_pearson = corr_matrix[0, 1]

scipy_corr, _ = pearsonr(x.flatten(), y)

model = LinearRegression()
model.fit(x, y)
y_pred = model.predict(x)

mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

slope = model.coef_[0]
intercept = model.intercept_

print("Karl Pearson’s Correlation Coefficient :", round(karl_pearson, 4))
print("Verified using scipy:", round(scipy_corr, 4))
print("Regression Equation: y =", round(slope, 2), "x +", round(intercept, 2))
print("Mean Squared Error (MSE):", round(mse, 4))
print("R² Score:", round(r2, 4))

plt.scatter(x, y, label="Data Points")
plt.plot(x, y_pred, color='red', linewidth=2, label="Regression Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression with Strong Correlation")
plt.legend()
plt.grid(True)
plt.show()
````

**1. What is correlation? How is it different from regression?**

Correlation is a statistical measure used to determine the strength and direction of the relationship between two variables. It indicates whether the variables increase together (positive correlation), decrease together (negative correlation), or have no relationship (zero correlation). The value of correlation lies between -1 and +1, where +1 indicates perfect positive correlation and -1 indicates perfect negative correlation.

Regression, on the other hand, is a statistical technique used to establish a functional relationship between variables and to predict the value of one variable based on another. It involves identifying a dependent variable (output) and an independent variable (input), and then fitting a mathematical model (usually a line).

The key difference is that correlation only measures the degree of association between variables, while regression is used for prediction and modeling. Correlation is symmetric (no distinction between variables), whereas regression is directional and defines dependent and independent variables.

---



**2. What is simple linear regression? Explain with suitable example.**

Simple linear regression is a statistical method used to model the relationship between one independent variable and one dependent variable using a straight-line equation:

            y=mx+c

where y is the dependent variable, x is the independent variable, m is the slope, and c is the intercept.

The main purpose of simple linear regression is to predict the value of the dependent variable based on the independent variable.

Example:
Consider a scenario where we want to predict a student’s marks based on the number of hours studied. If the data shows that as study hours increase, marks also increase, we can use linear regression to draw a best-fit line and predict marks for any given number of study hours.

**3. How do you calculate Pearson correlation in Python?**

Pearson correlation coefficient in Python can be calculated using the pearsonr() function from the SciPy library. This function returns both the correlation coefficient and the p-value.

Example code:

```
from scipy.stats import pearsonr
corr, _ = pearsonr(x, y)
print("Correlation:", corr)
```

This method computes the strength and direction of the linear relationship between two variables efficiently and accurately.

**4. What is multicollinearity? Does it affect simple regression?**

Multicollinearity is a situation in which two or more independent variables in a regression model are highly correlated with each other. This means that one variable can be predicted from another, which creates redundancy in the model.

Multicollinearity mainly affects multiple regression models, where more than one independent variable is present. It can lead to unstable coefficients, difficulty in interpreting the model, and reduced reliability.

However, it does not affect simple linear regression, because simple regression involves only one independent variable, so there is no possibility of correlation among multiple predictors.

**5. A dataset shows strong correlation but poor prediction accuracy. Why?**

A dataset may show strong correlation but still have poor prediction accuracy due to several reasons:

Correlation does not imply causation: Even if two variables are strongly related, one may not directly influence the other.
Presence of noise or outliers: Extreme values or random variations can affect model performance.
Non-linear relationship: Pearson correlation measures only linear relationships; if the actual relationship is non-linear, prediction will be poor.
Overfitting or underfitting: The model may either be too complex (overfitting) or too simple (underfitting), leading to inaccurate predictions.
Insufficient or poor-quality data: Limited or biased data can reduce the reliability of predictions.

Thus, strong correlation alone is not sufficient to guarantee a good predictive model.
